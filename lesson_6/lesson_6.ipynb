### Zero Check Decorator

Write a decorator function called `check` that verifies that the denominator is not equal to 0 and apply it to the following function:

python
@check
def div(a, b):
   return a / b



input: div(6, 2)
output: 3



input: div(6, 0)
output: "Denominator can't be zero"

```python
def check(func):
    def dec(a, b):
        try:
            return func(a, b)
        except ZeroDivisionError:
            print("Denominator can't be zero")

    return dec
    
@check
def div(a,b):
    return a/b

div(6,0)
```

**Objective**: Create a program to manage employee records using file handling.  

**Tasks**:  
1. **File Creation and Data Entry**  
   - Create a file named **"employees.txt"**.  
   - Allow the user to add new employee records. Each record should have the following fields:  
     

     Employee ID, Name, Position, Salary

```python
employees = {}
with open('employees.txt', "r") as records:
    for line in records:
        line = line.strip()
        if not line:
            continue
        parts = line.split(", ")
        employees[parts[0]] = parts[1:]

while True:
    print("Choose an option: \n 1)Add new employee record \n 2)View all employee records \n 3)Search an employee by Employee ID \n 4)Update an employee's information \n 5)Delete an employee record \n 6)Exit")
    ask = input("Choose one of the numbers: ")

    if ask == "1":
        emp_id = input("Enter Employee ID: ")
        name = input("Enter Name: ")
        position = input("Enter Position: ")
        salary = input("Enter Salary: ")
        new = [emp_id, name, position, salary]

        employees[emp_id] = [name, position, salary]

        with open("employees.txt", "a") as f1:
            f1.write(", ".join(new) + "\n")

    elif ask == "2":
        with open("employees.txt") as f1:
            for line in f1:
                print(line.strip())

    elif ask == "3":
        search = input("Enter ID of an employee: ")
        if search in employees:
            info = employees[search]
            print(f"ID: {search}, Name: {info[0]}, Position: {info[1]}, Salary: {info[2]}")
        else:
            print("No such employee")

    elif ask == "4":
        search = input("Enter ID of an employee: ")
        if search in employees:
            info = employees[search]
            asking = input("What do you want to update? : \n 1)name \n 2)position \n 3)salary\n")
            if asking == "1":
                info[0] = input("Enter new name: ")
            elif asking == "2":
                info[1] = input("Enter new position: ")
            elif asking == "3":
                info[2] = input("Enter new salary: ")

            with open('employees.txt', "w") as f:
                for emidd, data in employees.items():
                    f.write(", ".join([emidd] + data) + "\n")
            print("Employee data updated")
        else:
            print("No such employee")

    elif ask == "5":
        ask1 = input("Select an Employee by his ID to delete record: ")
        if ask1 in employees:
            del employees[ask1]
            print("Record successfully deleted!")

            with open("employees.txt", "w") as f:
                for emidd, data in employees.items():
                    f.write(", ".join([emidd] + data) + "\n")
        else:
            print("There is no such record")

    elif ask == "6":
        break



                              
                               




               

```

### **Word Frequency Counter**
**Objective**: Analyze a text file and count how often each word appears.  

**Tasks**:  
1. **File Input**  
   - Use the file **"sample.txt"**. The file can contain any text (like a paragraph or an article).  
   - If **"sample.txt"** does not exist, prompt the user to create it by typing in a paragraph.  

2. **Count Word Frequency**  
   - Read the file content and split it into individual words.  
   - Count the frequency of each word (ignore capitalization, e.g., "The" and "the" should be counted as the same word).  
   - Ignore punctuation (like commas, periods, etc.).  

3. **Output**  
   - Display the total number of words in the file.  
   - Display the top 5 most common words with their counts.  
   - Save the output to a new file called **"word_count_report.txt"**.  

4. **Example Output**  
   **Content of sample.txt**:  
   

   This is a simple file.
   This file, is for testing purposes. It is a test file.
   


   **Console Output**:  
   

   Total words: 14
   Top 5 most common words:
   is - 3 times
   this - 2 times
   file - 3 times
   a - 2 times
   test - 1 time
   


   **Content of word_count_report.txt**:  
   

   Word Count Report
   Total Words: 14
   Top 5 Words:
   is - 3
   file - 3
   this - 2
   a - 2
   test - 1
   

```python
#Q1 - file input
try:
    with open("sample.txt") as sample:
        read = sample.read()
except FileNotFoundError as e:
    prompt = input("sample.txt not found. Create it by texting a paragraph:")
    with open("sample.txt", "w") as f:
        f.write(prompt)
    read = prompt
#Q2 - count word frequency
import string
txt = ""
for char in read:
     if char not in string.punctuation:
          txt += char
txt = txt.lower()

words = txt.split()

count = {}
for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

#Q3 Output
total_length = len(words)
print(f"Total words: {total_length}")

sorted_words = sorted(count.items(), key= lambda item: item[1] , reverse= True)
top_5 = sorted_words[:5]
with open("word_count_report.txt", "w") as f:
    f.write("Word Count Report\n")
    f.write(f"Total Words: {total_length}\n")
    f.write("Top 5 Words:\n")
    for word, count in top_5:
        f.write(f"{word} - {count}\n")



```
