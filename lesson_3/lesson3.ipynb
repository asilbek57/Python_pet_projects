```python
#Q12
my_dict = {"a": 1, "b": 2, "c": 1, "d": 1}
target_value = 1
count = 0
for v in my_dict.values():
    if v == target_value:
        count = count + 1
print(count)
```

```python
#Q13
my_dict = {"a": 1, "b": 2, "c": 3}
inverted = {}
for key, value in my_dict.items():
    inverted[value] = key
print(inverted)
```

```python
#Q14
my_dict = {"a": 1, "b": 2, "c": 1}
target_value = 1
matching_keys = []
for key, value in my_dict.items():
    if value == target_value:
        matching_keys.append(key)
print(matching_keys)
```

```python
#Q15
keys_list = ["a", "b", "c"]
values_list = [1, 2, 3]
new_dict = dict(zip(keys_list, values_list))
print(new_dict)
```

```python
#Q16
my_dict = {"a": 1, "b": {"x": 10}}
has_nested = False
for value in my_dict.values():
    if isinstance(value, dict):
        has_nested = True
print(has_nested)
```

```python
#Q20
my_dict = {"c": 3, "a": 1, "b": 2}
sorted_by_key = dict(sorted(my_dict.items()))
print(sorted_by_key)
```

```python
#Q21
my_dict = {"a": 3, "b": 1, "c": 2}
sorted_by_value = dict(sorted(my_dict.items(), key=lambda item: item[1]))
print(sorted_by_value)
```

```python
#Q22
my_dict = {"a": 1, "b": 5, "c": 10, "d": 3}
filtered = {}
for key, value in my_dict.items():
    if value > 3:
        filtered[key] = value
print(filtered)
```
