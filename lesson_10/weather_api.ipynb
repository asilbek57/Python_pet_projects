## Task 1: Weather API
   1. Use this url : https://openweathermap.org/
   2. Use the `requests` library to fetch weather data for a specific city(ex. your hometown: Tashkent) and print relevant information (temperature, humidity, etc.).

```python
import requests
city = input("Enter City Name")
url = (f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=bab9632b78a8d0e267576ac68515ba3f&units=metric")

responce = requests.get(url)

if responce.status_code == 200:
    data = responce.json()
    
    temparature=data["main"]["temp"]
    humidity = data["main"]["humidity"]
    wind = data["wind"]
    city_name = data["name"]
    description = data["weather"][0]["description"]
    print(f"""Temparature in {city_name} is {temparature} °C
Humidity : {humidity}%
Wind: {wind}
Description for today: {description}

    """)
else:
    print(f"Error! Status code : {responce.status_code}")





```
